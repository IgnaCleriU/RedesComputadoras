# TP7 — Decodificar un mensaje de unos y ceros

Traduce cada grupo de **7 bits** a un carácter ASCII (por ejemplo, `1000001` → `A`). Cada línea se decodifica por separado.

Peguen las filas de unos y ceros entre las comillas triples y ejecuten la celda (`Shift+Enter`).

```python
TEXTO = """
1001000110111111011001100001
0100000110110111101011101110
1100100110111101000010100000
"""

for bits in TEXTO.split():
    print("".join(chr(int(bits[k:k + 7], 2)) for k in range(0, len(bits), 7)))
```
