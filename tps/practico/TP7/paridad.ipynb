# TP7 — Paridad de filas y columnas

Muestra la matriz con la paridad par de cada fila (a la derecha) y de cada columna (abajo).

Peguen las filas de unos y ceros entre las comillas triples y ejecuten la celda (`Shift+Enter`).

```python
TEXTO = """
1001000110111111011001100001
0100000110110111101011101110
1100100110111101000010100000
"""

filas = TEXTO.split()
for fila in filas:
    print(fila, "|" + str(fila.count("1") % 2))
print("-" * (len(filas[0]) + 2))
print("".join(str(sum(fila[j] == "1" for fila in filas) % 2) for j in range(len(filas[0]))))
```
